## Project Title:
Hallmark-Driven Gene Expression Signatures of Angiogenesis and Growth Suppression Evasion in Prostate and Ovarian Cancer. 


## Project Goal:
This project aims to investigate whether cancer hallmarks, specifically sustained angiogenesis and evading growth suppression, are predictive of patient survival and disease characteristics in aggressive, metastasis-prone cancers. We focus on prostate adenocarcinoma (PRAD) and ovarian serous carcinoma (OV) to explore how hallmark-associated gene expression relates to survival outcomes, tumor stage, and treatment response, and whether hallmark expression patterns differ across cancer types.


## Disease Background:

Cancer progression is driven by a set of conserved biological processes known as the hallmarks of cancer, which represent coordinated molecular programs that enable tumor growth, survival, and dissemination. This study focuses on two hallmarks that are particularly relevant to aggressive disease behavior: sustained angiogenesis and evasion of growth suppression.


Sustained angiogenesis enables tumors to grow beyond diffusion limits by inducing the formation of new blood vessels. Tumors hijack normal vascular signaling pathways to increase oxygen and nutrient supply, facilitating rapid growth and metastatic spread. Key genes involved in angiogenic signaling include VEGFA, FGF2, ANGPT2, HIF1A, and TNF. Angiogenesis plays a critical role in highly aggressive and metastatic cancers, particularly ovarian serous carcinoma, which relies heavily on angiogenic signaling to support diffuse peritoneal spread.


Evasion of growth suppression allows cancer cells to bypass normal regulatory mechanisms that limit cell proliferation. This hallmark is commonly achieved through disruption of tumor suppressor pathways governing cell cycle control, DNA damage response, and apoptosis. Frequently altered genes in these pathways include TP53, RB1, PTEN, ARID1A, CDKN2A (p16), and RUNX3. Loss of growth suppressive control is a defining feature of many advanced cancers and contributes to treatment resistance and unchecked tumor expansion.


This analysis focuses on prostate adenocarcinoma (PRAD) and ovarian serous carcinoma (OV), two cancers with distinct biological and clinical characteristics. PRAD is a common malignancy with highly variable outcomes, often progressing slowly but becoming aggressive in cases involving tumor suppressor loss such as PTEN, TP53, or RB1 alterations. In contrast, OV is one of the deadliest gynecologic cancers due to late-stage diagnosis and rapid metastatic spread, characterized by near-universal TP53 disruption and strong dependence on angiogenic pathways. These biological differences make PRAD and OV a compelling comparison for investigating whether hallmark-driven gene expression patterns capture meaningful distinctions in tumor behavior.


Because cancer hallmarks represent coordinated transcriptional programs rather than isolated genes, aggregating gene expression into hallmark-level scores provides a biologically motivated and interpretable feature space for machine learning analyses. This approach reduces dimensionality while preserving functional relevance, enabling downstream analyses to focus on dominant biological signals rather than genome-wide noise.

## Data-Set: 

For this analysis, we used gene expression and clinical metadata from the GSE62944 TCGA training dataset,focusing on Prostate Adenocarcinoma (PRAD) and Ovarian Serous Cystadenocarcinoma (OV)- one known to typically metastasize slowly, although could be aggressive in some cases, whereas the other is known to metastasize aggressively (Regional Cancer Care Associates, 2025; The University of Kansas Cancer Center, 2020). Our analysis centers on two specific hallmarks of cancer: angiogenesis and the evasion of growth suppressors.

In the gene expression data, each row corresponds to a gene (e.g., VEGFA, TP53), and each column represents an individual tumor sample. Expression values are reported as log₂-transformed TPM, which provides a normalized measure of mRNA abundance across each samples. The accompanying metadata links each sample to clinical characteristics such as cancer type, age, and sex. All data were generated through the TCGA project, where tumor samples were sequenced using RNA-seq. Raw read counts were normalized to TPM and then log₂-transformed to reduce variance and allow for more consistent comparison of gene expression patterns across the two cancer types.

## Data Analysis: 

### Methods


The machine learning techniques used in this analysis are Principal Component Analysis (PCA) and K-Means clustering, which are both unsupervised learning methods.PCA identifies directions that maximize the variance in the dataset, while K-Means clustering is used to group similar samples together based on feature similarity. These methods were used to explore gene expression patterns from cancer hallmark genes associated with: SUSTAINED ANGIOGENESIS & EVADING GROWTH SUPPRESSORS

These hallmarks are biologically important because they are linked to tumor development and cancer progression. By focusing on these pathways, the analysis aims to determine whether PRAD and OV samples show distinct expression patterns.

### Analysis

The gene expression data and metadata were first loaded and subsetted to include only PRAD and OV cancer samples. Hallmark gene sets for SUSTAINED ANGIOGENESIS and EVADING GROWTH SUPPRESSORS were selected because these pathways are important in tumor development. Only genes present in both the hallmark lists and the dataset were kept, which helped ensure the data were clean and consistent. Average expression values were then calculated for each hallmark to create angiogenesis and growth suppression scores for every sample.



PCA was used to explore patterns in the data after standardizing gene expression values. The PCA plot colored by cancer type showed clear separation between PRAD and OV samples, suggesting that these hallmark genes capture meaningful biological differences between the two cancers. 



When colored by angiogenesis score, the PCA plot showed gradual changes in score values across the plot, indicating that angiogenesis activity contributes to variation among samples. Similarly, when colored by growth suppression score, the PCA plot showed shifts in score values across the PCA space, suggesting that growth suppression activity also contributes to differences between samples.





K-Means clustering was applied using two clusters to identify natural groupings in the data. The resulting clusters closely matched the separation seen between PRAD and OV samples in the PCA plot. Because clustering was performed without using cancer type labels, this suggests that the observed groupings reflect real differences in gene expression. Overall, these results suggest that the selected hallmark genes contain useful information and support moving forward with a supervised learning model to classify cancer types.



## Verify and validate your analysis: 

Verify and Validate the Analysis

To verify and validate our analysis, we evaluated whether the structure identified by unsupervised learning methods was consistent, biologically interpretable, and supported by known cancer biology. Because this phase of the project relies on unsupervised methods such as Principal Component Analysis (PCA) and K‑means clustering, validation focused on the coherence of observed patterns rather than traditional predictive accuracy metrics.

Verification was performed by examining PCA projections colored by biologically relevant continuous variables. When samples were colored by angiogenesis score, a clear gradient was observed across PCA space, with higher angiogenesis scores clustering in distinct regions along the principal components. A similar pattern was observed when samples were colored by growth suppression scores, where gradual shifts in score values aligned with the dominant axes of variation. These continuous gradients indicate that the principal components capture variation related to known cancer hallmark activity rather than noise.

Additional validation was provided through K‑means clustering, which was performed without using cancer type labels. The resulting clusters closely aligned with the separation observed between prostate adenocarcinoma (PRAD) and ovarian serous carcinoma (OV) samples in PCA space. Because cancer type information was not used during clustering, this agreement suggests that the identified groupings reflect underlying differences in hallmark driven gene expression. Finally, the observed separation between PRAD and OV samples is consistent with existing literature showing that these cancers differ substantially in angiogenic signaling and tumor suppressor pathway disruption. Together, these observations support the conclusion that the selected hallmark gene sets capture meaningful biological structure and justify future supervised modeling to formally assess predictive performance.

To further validate the patterns observed in this analysis, we compared our results to prior work describing cancer hallmarks as coordinated transcriptional programs. The continuous gradients observed in PCA space when samples are colored by angiogenesis and growth suppression scores are consistent with the framework proposed by Hanahan, which emphasizes that cancer hallmarks operate at the pathway level and vary systematically across cancer types rather than acting through isolated genes. In particular, angiogenesis and evasion of growth suppression are described as central organizing processes that shape tumor behavior and progression. The fact that these hallmark scores align with the dominant axes of variation in PCA space, and that unsupervised K‑means clustering recapitulates known separation between prostate adenocarcinoma and ovarian serous carcinoma samples, supports the interpretation that the observed structure reflects meaningful biological differences rather than technical artifacts. Together, the agreement between unsupervised clustering, hallmark score gradients, and established cancer biology provides validation for the use of hallmark driven gene expression patterns in downstream supervised modeling.


Supervised Model Evaluation

To quantitatively evaluate how well hallmark activity can distinguish PRAD from OV tumors, we trained a logistic regression classifier using two biologically interpretable features: angiogenesis score and growth suppression score. We selected out-of-sample classification error (validation error as the primary metric because it directly measures how well the model generalizes to new, unseen data.

Out-of-sample error is calculated as:

[
Error = 1 - Accuracy
]

where accuracy represents the fraction of correctly classified samples. This metric is particularly useful because it reflects real-world model performance rather than performance on the training data used to fit the model.

The model achieved:

* **In-sample (training) error:** 36.3%
* **Out-of-sample (validation) error:** 39.4%



The validation accuracy of approximately 60.6% indicates that the model performs substantially better than random guessing (50% accuracy for a binary classifier). Importantly, the training error is slightly lower than the validation error by 3.1%, indicating that the model is performing well. However,given the high error rates across both datasets, the model is likely underfit, a result of its streamlined design and reliance on only two hallmark-derived features. This indicates that the identified patterns represent a genuine, generalizable biological signal rather than an overfit response to random fluctuations in the data.

To further assess model performance independent of a specific classification threshold, we examined the Receiver Operating Characteristic (ROC) curve and computed the Area Under the Curve (AUC).The model achieved an AUC of 0.603. This means the classifier ranks a randomly chosen OV sample higher than a randomly chosen PRAD sample about 60% of the time. Because an AUC of 0.50 corresponds to random performance, this result indicates meaningful predictive signal derived from hallmark activity.

## Conclusions and Ethical Implications: 

Conclusions

This study used hallmark-driven gene expression profiles to explore biological differences between prostate adenocarcinoma (PRAD) and ovarian serous carcinoma (OV). By focusing on genes associated with sustained angiogenesis and evasion of growth suppression, the analysis emphasized biologically meaningful features rather than genome-wide expression alone. Principal Component Analysis revealed clear separation between PRAD and OV samples in low-dimensional space, indicating that these hallmark pathways capture dominant sources of variation between the two cancer types. Unsupervised clustering using K-means further supported this distinction, with cluster assignments closely matching cancer type despite no labels being used during training.


Examination of PCA projections colored by hallmark scores showed smooth gradients of angiogenesis and growth suppression activity across the principal components. This alignment suggests that the major axes of variation correspond to biologically relevant hallmark activity rather than technical noise. Together, the unsupervised results demonstrate that angiogenesis and growth suppression pathways play distinct and measurable roles in structuring gene expression patterns across cancers with different metastatic behaviors.


Building on these findings, a supervised logistic regression classifier trained on angiogenesis and growth suppression scores was able to distinguish PRAD from OV with approximately 60 percent validation accuracy and an AUC of approximately 0.60. While this performance is modest, it consistently exceeded random classification, indicating the presence of a reproducible biological signal. The similarity between training and validation error suggests that the model did not overfit, but instead underfit due to its reliance on only two hallmark-derived features. These results indicate that while angiogenesis and growth suppression contain meaningful discriminatory information, they are insufficient on their own to fully capture the complexity of cancer type classification.

The class trends between the angiogenesis and growth suppression scores exhibit a consistent linear relationship across both PRAD and OV, suggesting that these hallmark processes are interconnected and shaped by a shared underlying structure rather than varying independently between cancer types. This pattern is also reflected in all PCA prohections, where neither group forms a clearly independent or opposing trajectory. This reinforces the idea that the observed variation is driven more by coordinated biological regulation than by distinct, cancer-type-specific behavior.

Overall, this work demonstrates that biologically informed dimensionality reduction and feature engineering can reveal meaningful cancer-specific structure in gene expression data. The consistency between unsupervised and supervised analyses supports the conclusion that hallmark-level transcriptional programs provide a valuable foundation for future modeling efforts that incorporate additional biological pathways and clinical context.


Ethical Implications


The use of gene expression based machine learning approaches in cancer research raises important ethical considerations. Public genomic datasets such as TCGA may reflect underlying demographic, socioeconomic, and healthcare access biases, which can limit the generalizability of findings to more diverse patient populations. Models trained on such data risk encoding and perpetuating these biases if applied without careful consideration.


There is also a risk of overinterpreting exploratory or correlational findings, particularly when unsupervised methods are used. While the patterns observed in this study are biologically consistent and reproducible, they do not establish causality or clinical utility. Applying underperforming or incompletely validated models in clinical settings could lead to misclassification or inappropriate conclusions, with potential downstream consequences for patient care.


Although TCGA data are de-identified, the broader use of genomic information raises ongoing concerns related to patient privacy, data security, and informed consent. Ethical application of these methods requires transparency regarding model limitations, careful validation across independent cohorts, and clear communication that results are hypothesis-generating rather than clinically actionable. Responsible interpretation is particularly important when simplifying complex biological systems into reduced feature representations such as hallmark scores.

## Limitations and Future Work: 

Limitations

Several limitations should be considered when interpreting the results of this analysis. First, the study examines only two cancer types, which restricts the ability to generalize findings across a broader range of malignancies. Second, the analysis focuses on two predefined hallmark gene sets, which improves interpretability but may exclude other important molecular processes contributing to tumor behavior. Limiting the feature space to angiogenesis and growth suppression likely contributed to underfitting in the supervised model.


From a data perspective, TCGA samples may not fully represent the diversity of patient populations, disease stages, or treatment histories seen in clinical practice. Additionally, although standard normalization procedures were applied, technical variation inherent to RNA sequencing data may still influence results.


From a modeling standpoint, the use of linear classifiers and a small number of features limits the complexity of relationships the model can capture. Unsupervised methods reveal structure in the data but do not directly assess clinical relevance or predictive utility. Finally, clinical variables such as tumor stage, treatment response, and survival outcomes were not incorporated, preventing direct linkage between gene expression patterns and patient-level outcomes.


Future Work


Future work will extend this exploratory analysis by explicitly addressing the observed underfitting and expanding the biological feature space. Incorporating additional cancer hallmarks such as epithelial-mesenchymal transition, hypoxia, immune evasion, and DNA damage response may capture complementary pathways that improve predictive performance. Expanding beyond two hallmark-derived features is likely to enhance model expressiveness while maintaining biological interpretability.


More rigorous supervised learning approaches, including regularized logistic regression, support vector machines, or ensemble methods, can be employed alongside stratified cross-validation to better estimate out-of-sample performance. Regression-based models could be used to investigate relationships between hallmark activity and survival outcomes.


Future analyses should also integrate clinical covariates and evaluate performance across cancer stages and subtypes. Validation using independent external datasets will be essential for assessing robustness and generalizability. Together, these extensions would strengthen the translational relevance of hallmark-driven gene expression modeling while preserving its interpretability and biological grounding.
